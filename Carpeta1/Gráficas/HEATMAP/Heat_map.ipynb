```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)

# ============================================
# 1. CREAR 50 ORGANISMOS
# ============================================

n_moscas = 50

matriz = pd.DataFrame({
    "Longitud_cuerpo": np.random.normal(5.0, 0.5, n_moscas),
    "Longitud_alas": np.random.normal(3.2, 0.3, n_moscas),
    "Peso": np.random.normal(0.85, 0.10, n_moscas),
    "Velocidad_vuelo": np.random.normal(32, 5, n_moscas),
    "Supervivencia": np.random.normal(28, 6, n_moscas)
})

# Evitar valores negativos
matriz = matriz.clip(lower=0)

# Identificador de cada organismo
matriz.insert(
    0,
    "Organismo",
    [f"Mosca_{i+1}" for i in range(n_moscas)]
)


# ============================================
# 4. MAPA DE CALOR
# ============================================

plt.figure(figsize=(10, 8))

sns.heatmap(
    matriz.iloc[:, 1:],
    cmap="viridis",
    yticklabels=matriz["Organismo"],
    annot=False
)

plt.title("Mapa de características de 50 moscas")
plt.xlabel("Características")
plt.ylabel("Organismos")

plt.tight_layout()

plt.savefig(
    "mapa_calor_50_moscas.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
