```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)

# ============================================
# 1. CREAR 50 ORGANISMOS
# ============================================

n_moscas = 50

matriz = pd.DataFrame({
    "Longitud_cuerpo": np.random.normal(5.0, 0.5, n_moscas),
    "Longitud_alas": np.random.normal(3.2, 0.3, n_moscas),
    "Peso": np.random.normal(0.85, 0.10, n_moscas),
    "Velocidad_vuelo": np.random.normal(32, 5, n_moscas),
    "Supervivencia": np.random.normal(28, 6, n_moscas)
})

# Evitar valores negativos
matriz = matriz.clip(lower=0)

# Identificador de cada organismo
matriz.insert(
    0,
    "Organismo",
    [f"Mosca_{i+1}" for i in range(n_moscas)]
)

# ============================================
# 2. MOSTRAR LA MATRIZ
# ============================================

display(matriz.round(2))
```
